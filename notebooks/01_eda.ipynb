```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

RAW_DIR = '../data/raw'
os.listdir(RAW_DIR)
```

## 1. Mandi Master Data Cleaning
Standardizing mandi_id using regex.

```python
df_mandi = pd.read_csv(os.path.join(RAW_DIR, 'track3_mandi_master.csv'))
print('Raw unique IDs:', df_mandi['mandi_id'].nunique())

# Cleaning logic
import re
def clean_id(m_id):
    if pd.isna(m_id): return np.nan
    clean = re.sub(r'[^A-Z0-9]', '', str(m_id).upper())
    if clean.startswith('MANDI'): return clean
    if clean.startswith('M'): return 'MANDI' + clean[1:].zfill(3)
    if clean.isdigit(): return 'MANDI' + clean.zfill(3)
    return clean

df_mandi['mandi_id_clean'] = df_mandi['mandi_id'].apply(clean_id)
print('Clean unique IDs:', df_mandi['mandi_id_clean'].nunique())
```
